```python
import io
import json
import random
import zipfile
from pathlib import Path

import requests
import torch
import torch.nn.functional as F
import transformers
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from transformers import AutoProcessor, CLIPModel
import numpy as np
from concurrent.futures import ThreadPoolExecutor, as_completed
from tqdm.auto import tqdm
```

```python
ROOT = Path.cwd()

DATA = ROOT / "data"
IMAGES = DATA / "images"

PROJECT = Path("/content/drive/MyDrive/MLLM/")
OUTPUTS = PROJECT / "outputs"

for folder in (DATA, IMAGES, OUTPUTS):
    folder.mkdir(parents=True, exist_ok=True)

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Folder:", ROOT)
print("Device:", device)
print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
```

```python
manifest_path = DATA / "xm3600_subset500.json"

if not manifest_path.exists():
    url = "https://google.github.io/crossmodal-3600/web-data/captions.zip"

    response = requests.get(url, timeout=60)
    response.raise_for_status()

    eligible = []

    with zipfile.ZipFile(io.BytesIO(response.content)) as archive:
        with archive.open("captions.jsonl") as source:
            for line in source:
                item = json.loads(line)

                en = item.get("en", {}).get("caption", [])
                bn = item.get("bn", {}).get("caption", [])

                if en and bn:
                    eligible.append({
                        "image_id": item["image/key"],
                        "en": en,
                        "bn": bn,
                    })

    eligible.sort(key=lambda item: item["image_id"])
    subset = random.Random(42).sample(eligible, 500)

    manifest_path.write_text(
        json.dumps(subset, ensure_ascii=False, indent=2),
        encoding="utf-8"
    )

subset = json.loads(manifest_path.read_text(encoding="utf-8"))
print("Fixed image count:", len(subset))
```

```python
demo = subset[:3]
image_ids = [item["image_id"] for item in demo]
captions = [item["en"][0] for item in demo]

images = []

for image_id in image_ids:
    path = IMAGES / f"{image_id}.jpg"

    if not path.exists():
        url = (
            "https://open-images-dataset.s3.amazonaws.com/"
            f"crossmodal-3600/{image_id}.jpg"
        )
        response = requests.get(url, timeout=60)
        response.raise_for_status()
        path.write_bytes(response.content)

    with Image.open(path) as image:
        images.append(image.convert("RGB"))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

for i, (axis, image, caption) in enumerate(zip(axes, images, captions)):
    axis.imshow(image)
    axis.set_title(f"I{i}")
    axis.axis("off")
    print(f"I{i} / T{i}: {caption}")

plt.tight_layout()
plt.show()
```

**Model and embeddings**

```python
MODEL_ID = "openai/clip-vit-base-patch16"

processor = AutoProcessor.from_pretrained(MODEL_ID)
model = CLIPModel.from_pretrained(MODEL_ID).to(device).eval()
```

```python
batch = processor(
    images=images,
    text=captions,
    padding=True,
    truncation=True,
    max_length=77,
    return_tensors="pt"
).to(device)

with torch.inference_mode():
    outputs = model(**batch)
```

```python
image_embeddings = outputs.image_embeds.float().cpu()
text_embeddings = outputs.text_embeds.float().cpu()

print("Image embeddings:", image_embeddings.shape)
print("Text embeddings:", text_embeddings.shape)
print("Image vector lengths:", image_embeddings.norm(dim=-1))
print("Text vector lengths:", text_embeddings.norm(dim=-1))
```

```python
example = torch.tensor([[3.0, 4.0]])
print("Normalized [3, 4]:", F.normalize(example, dim=-1))
```

**Similarity matrix**

```python
similarity = image_embeddings @ text_embeddings.T

similarity_df = pd.DataFrame(
    similarity.numpy(),
    index=["I0", "I1", "I2"],
    columns=["T0", "T1", "T2"],
)

display(similarity_df.round(3))

for i in range(3):
    best_j = similarity[i].argmax().item()
    print(f"I{i} → T{best_j}: {captions[best_j]}")
```

**Contrastive learning and temperature**

```python
def symmetric_loss(scores, temperature):
    logits = scores / temperature
    targets = torch.arange(scores.shape[0])
    
    image_to_text = F.cross_entropy(logits, targets)
    text_to_image = F.cross_entropy(logits.T, targets)
    
    return (image_to_text + text_to_image) / 2
```

```python
toy_similarity = torch.tensor([
    [0.8, 0.2],
    [0.1, 0.7]
])

for temperature in (1.0, 0.1):
    correct = symmetric_loss(toy_similarity, temperature)
    
    wrong = symmetric_loss(
        toy_similarity[:, [1, 0]],
        temperature
    )
    
    print(
        f"\nTemperature={temperature}: "
        f"correct loss={correct.item():.4f}, "
        f"wrong loss={wrong.item():.4f}"
    )

    print("Row probabilities:")
    print((toy_similarity / temperature).softmax(dim=-1))
```

**Prediction using fixed class prompts**

```python
labels = ["a city", "food", "a flower", "a dog", "a car"]
prompts = [f"a photo of {label}." for label in labels]

batch = processor(
    images=images,
    text=prompts,
    padding=True,
    truncation=True,
    max_length=77,
    return_tensors="pt"
).to(device)
```

```python
with torch.inference_mode():
    zero_outputs = model(**batch)

logits = zero_outputs.logits_per_image.float().cpu()
relative_scores = logits.softmax(dim=-1)

zero_shot_df = pd.DataFrame(
    relative_scores.numpy(),
    index=["I0", "I1", "I2"],
    columns=labels
)

display(zero_shot_df.round(3))
```

```python
for i in range(3):
    predicted = relative_scores[i].argmax().item()
    print(
        f"I{i}: {labels[predicted]} "
        f"| relative score={relative_scores[i, predicted]:.3f}"
    )
```

```python
keep = [i for i, label in enumerate(labels) if label != "a flower"]

reduced_scores = logits[2, keep].softmax(dim=-1)
best = reduced_scores.argmax().item()

print("Flower image, with flower class removed:")
print("Prediction:", labels[keep[best]])
print("Relative score:", reduced_scores[best].item())
```

**Queries IDs**

```python
subset = json.loads(manifest_path.read_text(encoding="utf-8"))

eval_image_ids = [item["image_id"] for item in subset]
caption_records = [
    {
    "image_id" : item["image_id"],
    "caption_index" : j,
    "text" : caption
    }
    for item in subset
    for j, caption in enumerate(item["en"])
]

texts = [record["text"] for record in caption_records]
text_image_ids = [record["image_id"] for record in caption_records]


print("Images:", len(eval_image_ids))
print("English captions:", len(texts))
```

```python
def download_image(image_id):
    path = IMAGES / f"{image_id}.jpg"

    if path.exists():
        try:
            with Image.open(path) as image:
                image.verify()
            return
        except (OSError, SyntaxError):
            pass
    
    url = (
        "https://open-images-dataset.s3.amazonaws.com/"
        f"crossmodal-3600/{image_id}.jpg"
    )

    response = requests.get(url, timeout=(15, 60))
    response.raise_for_status()

    with Image.open(io.BytesIO(response.content)) as image:
        image.verify()

    temporary = path.with_suffix(".jpg.part")
    temporary.write_bytes(response.content)
    temporary.replace(path)
    

with ThreadPoolExecutor(max_workers=8) as executor:
    futures = [
        executor.submit(download_image, image_id)
        for image_id in eval_image_ids
    ]
    
    for future in tqdm(
        as_completed(futures),
        total=len(futures),
        desc="Images"
    ):
        future.result()
```

**Image and text embeddings**

```python
def load_rgb(image_id):
    with Image.open(IMAGES / f"{image_id}.jpg") as image:
        return image.convert("RGB")
    

def feature_tensor(result):
    if isinstance(result, torch.Tensor):
        return result
    
    return result.pooler_output
```

```python
MAX_TOKENS = model.config.text_config.max_position_embeddings

@torch.inference_mode()
def encode_clip(items, kind, batch_size):
    chunks = []
    
    for start in tqdm(
        range(0, len(items), batch_size),
        desc=f"CLIP {kind}"
    ):
        part = items[start:start + batch_size]
        if kind == "image":
            batch = processor(
                images=[load_rgb(image_id) for image_id in part],
                return_tensors="pt"
            ).to(device)
        
            result = model.get_image_features(
                **batch, return_dict=True
            )
        
        else:
            batch = processor(
                text=part,
                padding=True,
                truncation=True,
                max_length=MAX_TOKENS,
                return_tensors="pt",
            ).to(device)

            result = model.get_text_features(
                **batch, return_dict=True
            )
        
        features = feature_tensor(result).float()
        chunks.append(F.normalize(features, dim=-1).cpu())

    return torch.cat(chunks, dim=0)
```

```python
image_features = encode_clip(eval_image_ids, kind="image", batch_size=16)

text_features = encode_clip(texts, kind="text", batch_size=64)

token_ids = processor.tokenizer(
    texts, truncation=False, padding=False, verbose=False
)["input_ids"]

truncated_count = sum(len(ids) > MAX_TOKENS for ids in token_ids)

print("Image embeddings:", image_features.shape)
print("Text embeddings:", text_features.shape)
print("Captions exceeding token limit:", truncated_count)
```

**Similarity → ranking → ID matching → Recall**

```python
def recall_at_k(scores, query_ids, candidate_ids, ks=(1, 5, 10)):
    scores = np.asarray(scores)
    query_ids = np.asarray(query_ids)
    candidate_ids = np.asarray(candidate_ids)
    
    ranked = np.argsort(
        -scores, axis=1, kind="stable"
    )[:, :max(ks)]
    matches = candidate_ids[ranked] == query_ids[:, None]
    
    metrices = {
        f"R@{k}": 100 * matches[:, :k].any(axis=1).mean()
        for k in ks
    }
    
    return metrices, ranked
```

```python
S = (image_features @ text_features.T).numpy()

i2t_metrics, i2t_top = recall_at_k(
    S, eval_image_ids, text_image_ids
)

t2i_metrics, t2i_top = recall_at_k(
    S.T, text_image_ids, eval_image_ids
)

metrics_df = pd.DataFrame([
    {
        "model": "CLIP",
        "language": "en",
        "direction": "Image → Text",
        "queries": len(eval_image_ids),
        "candidates": len(texts),
        **i2t_metrics,
    },
    {
        "model": "CLIP",
        "language": "en",
        "direction": "Text → Image",
        "queries": len(texts),
        "candidates": len(eval_image_ids),
        **t2i_metrics,
    },
])

print("Similarity shape:", S.shape)
display(metrics_df.round(2))
```

**Text → Image top-5**

```python
q = 0

print("Query:", texts[q])
print("Correct image ID:", text_image_ids[q])

fig, axes = plt.subplots(1, 5, figsize=(16, 4))

for rank, (axis, image_index) in enumerate(
    zip(axes, t2i_top[q, :5]), start=1
):
    image_id = eval_image_ids[image_index]
    correct = image_id == text_image_ids[q]

    axis.imshow(load_rgb(image_id))
    axis.set_title(
        f"Rank {rank} | {S[image_index, q]:.3f}\n"
        f"{'MATCH' if correct else 'OTHER'}"
    )
    axis.axis("off")

plt.tight_layout()
plt.show()
```

****Image → Text top-5****

```python
q = 0

plt.figure(figsize=(4, 4))
plt.imshow(load_rgb(eval_image_ids[q]))
plt.axis("off")
plt.show()

print("Reference captions:", subset[q]["en"])

retrieved = []

for rank, text_index in enumerate(i2t_top[q, :5], start=1):
    retrieved.append({
        "rank": rank,
        "caption": texts[text_index],
        "similarity": float(S[q, text_index]),
        "correct": text_image_ids[text_index] == eval_image_ids[q],
    })

display(pd.DataFrame(retrieved).round(3))
```
